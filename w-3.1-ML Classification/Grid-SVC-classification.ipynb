```python
import pandas as pd
import torch
print("GPU:",torch.cuda.get_device_name(0))
```

```python
dataset = pd.read_csv("Social_Network_Ads.csv")
dataset
```

```python
dataset = pd.get_dummies(dataset, dtype=int, drop_first=True)
dataset
```

```python
dataset.columns
```

```python
indep = dataset[['Age', 'EstimatedSalary',  'Gender_Male']]
depend = dataset[['Purchased']]
```

```python
depend.value_counts()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(indep,depend,test_size=0.30,random_state=0)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC

svc_param_grid = {
    "C": [0.1, 1, 10, 100],
    "kernel": ["linear", "rbf", "poly"],
    "gamma": ["scale", "auto"],
    "degree": [2, 3, 4]   
}




grid = GridSearchCV(SVC(), svc_param_grid, refit=True ,n_jobs=3, scoring='f1_weighted')
grid.fit(X_train,y_train)
```

```python
re = grid.cv_results_
print("The classification value for best parameter {}:".format(grid.best_params_))
table = pd.DataFrame.from_dict(re)
table
```

```python
y_pred = grid.predict(X_test)
y_pred
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_pred,y_test)
print(cm)
```

```python
from sklearn.metrics import classification_report
report= classification_report(y_test, y_pred)
print(report)
```
