```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
dataset
```

```python
X = dataset.iloc[:, 3:5].values
X
```

```python
import scipy.cluster.hierarchy as sch
dendrogram = sch.dendrogram(sch.linkage(X,method="ward"))
plt.title('Dendrogram')
plt.xlabel('Customers')
plt.ylabel('Educlidean distance')
plt.show()
```

```python
from sklearn.cluster import AffinityPropagation
cusmodel = AffinityPropagation(affinity='euclidean',random_state = 5)
y = cusmodel.fit_predict(X)
```

```python
import numpy as np
cusmodel = AffinityPropagation(affinity='euclidean', random_state=5)
y = cusmodel.fit_predict(X)

cluster_centers_indices = cusmodel.cluster_centers_indices_
labels = cusmodel.labels_
n_clusters = len(cluster_centers_indices)

# Plot
plt.figure(figsize=(8, 6))

colors = plt.cm.tab10(np.linspace(0, 1, n_clusters))

for k, col in zip(range(n_clusters), colors):
    class_members = (labels == k)
    cluster_center = X[cluster_centers_indices[k]]
    
    # All points in cluster
    plt.scatter(X[class_members, 0], X[class_members, 1], color=col)
    
    # Cluster center
    plt.scatter(cluster_center[0], cluster_center[1], color='black', s=200, marker='x')
    
    # Join lines from center → each point
    for x in X[class_members]:
        plt.plot([cluster_center[0], x[0]], [cluster_center[1], x[1]], color=col, linewidth=1)

plt.title("Affinity Propagation Clustering")
plt.show()
```
